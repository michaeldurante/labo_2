
<h1 id="Introduction-to-Python-Classes">Introduction to Python Classes</h1><p>In Python, a class is a user-defined data type that contains both the data itself and the methods that may be used to manipulate it.</p>


```python
import numpy as np
import matplotlib.pyplot as plt 
import matplotlib.colors as colors
import pickle
from astropy.io import fits 
```


<p><a href="https://docs.python.org/3/tutorial/classes.html">Link to Python Class documentation</a></p>
<p>When to use a class:</p>
<ol>
<li>You want to define several functions to perform a specific task, and you want to use these functions in a self-consistent way repeatedly;</li>
<li>Your functions use shared parameters; you want each function to use the same values every time;</li>
<li>You want to use these functions and variables in different Jupyter notebooks without rewriting them every time and avoid dangerous copying and pasting.</li>
<li>If you spot an error in your data analysis, you only need to fix the error in one place and rerun all the analysis (also on a different Jupyter notebook) by just recalling the updated class.</li>
</ol>

<h2 id="Class-objects">Class objects</h2><p>Class objects support two kinds of operations: <em>attribute references</em> and <em>instantiation</em>.</p>
<p><strong>Attribute references</strong> use the standard syntax as for all attribute references in Python: <code>obj.name</code>. Let's take this example:</p>


```python
class MyClass:
    """A simple example class"""
    internal_variable = 12345

    def print_hello(self):
        return 'Hello World'
```


<p>You can retrieve the value of <code>internal_variable</code> and use the function <code>print_hello</code> directly by calling the class:</p>


```python
print(MyClass.internal_variable)
MyClass.print_hello(MyClass)
```


<p>Notice the argument <code>self</code> of the class: it means that the function requires the class itself as an argument. However, it is very rare to see a Class used in this way.</p>
<p><strong>Class instantiation</strong> uses function notation.
The example below creates a new instance of the class and assigns this object to the local variable <code>first_class</code>. The same action is repeated using the variable <code>second_class</code>.</p>


```python
first_class = MyClass()
second_class = MyClass()
```


<p>This action is called <em>class instantiation</em>. <br/>
We have created two Python objects, called <code>first_class</code> and <code>second_class</code>, which will have all the attributes and functions of the original class but will work independently. For example, we can change the value of the <code>internal_variable</code> of one of the two objects without affecting the other:</p>


```python
second_class.internal_variable = 67890
print('first_class internal variable value:', first_class.internal_variable)
print('second_class internal variable value:', second_class.internal_variable)
```


<p>A variable inside a class takes the name of <strong>data attribute</strong> or simply <strong>attribute</strong>
A function defined inside a class takes the name of <strong>method</strong>.</p>
<p>When calling a method of an instance object, you don't need to specify the <code>self</code> argument anymore, as the class will know to <em>take itself</em> as an argument:</p>


```python
print('running class method form the instance object:')
second_class.print_hello()
```


<h2 id="Class-initialization">Class initialization</h2><p>Classes support a special kind of method that is automatically called every time you create a new class instance. This <em>initialization</em> method is defined by the name <code>__init__</code></p>
<p>Let's analyze this example:</p>


```python


class MySecondClass:
    def __init__(self, x_input, y_input):
        self.x_pos = x_input
        self.y_pos = y_input


```


<p>When creating a new object using our class, the <code>__init__</code> method is automatically invoked. <br/>
In this example, <code>__init__</code> requires two arguments other than the <code>self</code> one, so we have to provide them when doing <em>class instantiation</em> to avoid an error:</p>


```python


# print('wrong class instantiation')
# new_class = MySecondClass()


```

```python


print('correct class instantiation (no output error)')
new_class = MySecondClass(258, 76)


```


<p>The two arguments <code>x_input, y_input</code> are now saved as attributes of the class <code>self.x_pos, self.y_pos</code>. Mind the difference:</p>
<ol>
<li>When defining a new method, you can recall these attributes using the <code>self.</code> appendix.</li>
<li>When calling the attributes from an instance object, you call them directly using the name of the object</li>
</ol>
<p>If you change the class, you must do a new instance, as in the example below.</p>


```python


class MySecondClass:
    def __init__(self, x_input, y_input):
        self.x_pos = x_input
        self.y_pos = y_input
    def print_position(self):
        print('attributes called from a class method', self.x_pos, self.y_pos)
    def change_position(self, x_new, y_new): 
        self.x_pos = x_new
        self.y_pos = y_new
        
new_class =  MySecondClass(258, 76)
new_class.print_position()

print('attributes called from instance object:', new_class.x_pos, new_class.y_pos )


```


<p>You can write a method to change one or more attributes.
In order to work, the method must get <code>self</code> as the argument, and the attributes to be changed must be referenced as <code>self.</code></p>

```python


new_class.change_position(789, 123)
new_class.print_position()


```

All the methods called after the change will use the updated value:</p>

<h3 id="Incorporating-the-centroid-algorithm">Incorporating the centroid algorithm</h3><ol>
<li>Take the algorithm to compute the centroid position (e.g., the one with the maximum number of iterations) and encapsulate it inside a method without forgetting the <code>self</code> argument;</li>
<li>You have to provide the science frame in input (as an argument), so you can repeat the calculation after subtracting the sky background;</li>
<li>You can provide the maximum number of iterations <code>maximum_number_of_iterations</code> as an argument, with a default value;</li>
<li>The method must provide in output the refined coordinates of the centre</li>
</ol>



<div class="highlight"><pre><span></span>    <span class="k">def</span><span class="w"> </span><span class="nf">compute_centroid</span><span class="p">(</span><span class="bp">self</span><span class="p">,</span> <span class="n">science_frame</span><span class="p">,</span> <span class="n">x_target_initial</span><span class="p">,</span> <span class="n">y_target_initial</span><span class="p">,</span> <span class="n">maximum_number_of_iterations</span><span class="o">=</span><span class="mi">20</span><span class="p">)</span>

        <span class="k">for</span> <span class="n">i_iter</span> <span class="ow">in</span> <span class="nb">range</span><span class="p">(</span><span class="mi">0</span><span class="p">,</span> <span class="n">maximum_number_of_iterations</span><span class="p">):</span>

            <span class="k">if</span> <span class="n">i_iter</span> <span class="o">==</span> <span class="mi">0</span><span class="p">:</span>
                <span class="c1"># first iteration</span>
                <span class="n">x_target_previous</span> <span class="o">=</span> <span class="n">x_target_initial</span>
                <span class="n">y_target_previous</span> <span class="o">=</span> <span class="n">y_target_initial</span>
            <span class="k">else</span><span class="p">:</span>
                <span class="c1"># using the previous result as starting point</span>
                <span class="n">x_target_previous</span> <span class="o">=</span> <span class="n">x_target_refined</span>
                <span class="n">y_target_previous</span> <span class="o">=</span> <span class="n">y_target_refined</span>

            <span class="c1"># 2D array with the distance of each pixel from the target star </span>
            <span class="n">target_distance</span> <span class="o">=</span> <span class="n">np</span><span class="o">.</span><span class="n">sqrt</span><span class="p">((</span><span class="bp">self</span><span class="o">.</span><span class="n">X</span><span class="o">-</span><span class="n">x_target_previous</span><span class="p">)</span><span class="o">**</span><span class="mi">2</span> <span class="o">+</span> <span class="p">(</span><span class="bp">self</span><span class="o">.</span><span class="n">Y</span><span class="o">-</span><span class="n">y_target_previous</span><span class="p">)</span><span class="o">**</span><span class="mi">2</span><span class="p">)</span>

            <span class="c1"># Selection of the pixels within the inner radius</span>
            <span class="n">annulus_sel</span> <span class="o">=</span> <span class="p">(</span><span class="n">target_distance</span> <span class="o">&lt;</span> <span class="bp">self</span><span class="o">.</span><span class="n">sky_inner_radius</span><span class="p">)</span>

            <span class="c1"># Weighted sum of coordinates</span>
            <span class="n">weighted_X</span> <span class="o">=</span> <span class="n">np</span><span class="o">.</span><span class="n">sum</span><span class="p">(</span><span class="n">science_frame</span><span class="p">[</span><span class="n">annulus_sel</span><span class="p">]</span><span class="o">*</span><span class="bp">self</span><span class="o">.</span><span class="n">X</span><span class="p">[</span><span class="n">annulus_sel</span><span class="p">])</span>
            <span class="n">weighted_Y</span> <span class="o">=</span> <span class="n">np</span><span class="o">.</span><span class="n">sum</span><span class="p">(</span><span class="n">science_frame</span><span class="p">[</span><span class="n">annulus_sel</span><span class="p">]</span><span class="o">*</span><span class="bp">self</span><span class="o">.</span><span class="n">Y</span><span class="p">[</span><span class="n">annulus_sel</span><span class="p">])</span>

            <span class="c1"># Sum of the weights</span>
            <span class="n">total_flux</span> <span class="o">=</span> <span class="n">np</span><span class="o">.</span><span class="n">sum</span><span class="p">(</span><span class="n">science_frame</span><span class="p">[</span><span class="n">annulus_sel</span><span class="p">])</span>

            <span class="c1"># Refined determination of coordinates</span>
            <span class="n">x_target_refined</span> <span class="o">=</span> <span class="n">weighted_X</span><span class="o">/</span><span class="n">total_flux</span>
            <span class="n">y_target_refined</span> <span class="o">=</span> <span class="n">weighted_Y</span><span class="o">/</span><span class="n">total_flux</span>

            <span class="n">difference_x</span> <span class="o">=</span> <span class="n">np</span><span class="o">.</span><span class="n">abs</span><span class="p">(</span><span class="n">x_target_refined</span><span class="o">-</span><span class="n">x_target_previous</span><span class="p">)</span> 
            <span class="n">difference_y</span> <span class="o">=</span> <span class="n">np</span><span class="o">.</span><span class="n">abs</span><span class="p">(</span><span class="n">y_target_refined</span><span class="o">-</span><span class="n">y_target_previous</span><span class="p">)</span>
            <span class="c1"># exit condition: difference between two consecutive iterations is smaller than 0.05 pixel for both</span>
            <span class="k">if</span> <span class="n">np</span><span class="o">.</span><span class="n">abs</span><span class="p">(</span><span class="n">difference_x</span><span class="p">)</span><span class="o">&lt;</span><span class="mf">0.1</span> <span class="ow">and</span>  <span class="n">np</span><span class="o">.</span><span class="n">abs</span><span class="p">(</span><span class="n">difference_y</span><span class="p">)</span><span class="o">&lt;</span><span class="mf">0.1</span><span class="p">:</span>
                  <span class="k">break</span>

        <span class="k">return</span> <span class="n">x_target_refined</span><span class="p">,</span> <span class="n">y_target_refined</span>
</pre></div>

<p><strong>HOMEWORK</strong> <br/>
You will need to <strong>convert JD to BJD_TDB</strong>!! You need to do it only for the JD array of the target star (the reference star will share the same time array), as we are interested in the precise timing of the exoplanetary signal. You can perform the conversion using the same script as in lecture 03, "Science Analysis."</p>


```python


from aperture_class import AperturePhotometry

from time import time
t0 = time()
target_star = AperturePhotometry()
target_star.provide_aperture_parameters(15, 25, 10, 347, 34)
target_star.aperture_photometry()
t1 = time()
print('elapsed_time=', t1-t0)


```
